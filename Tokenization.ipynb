```python
from nltk.tokenize import sent_tokenize

text="I like apples and prefer eating more fruits than any other food"
sent_tokenize(text)
```

```python
from nltk.tokenize import sent_tokenize

text="I like apples and prefer eating more fruits than any other food. But apple is always my favourite one"
sent_tokenize(text)
```

```python
from nltk.tokenize import word_tokenize

text="I like apples and prefer eating more fruits than any other food. But apple is always my favourite one"
word_tokenize(text)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = "running"

stemmed_words = stemmer.stem(words)
print(stemmed_words)
```
