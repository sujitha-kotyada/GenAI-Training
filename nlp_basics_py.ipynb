```python
from nltk.tokenize import sent_tokenize
text="Artificial Intelligence is transforming industries. NLP is a key part of AI. Tokenization is the first NLP step."
sent_tokenize(text)
```

```python
from nltk.tokenize import word_tokenize
text="Artificial Intelligence is transforming industries. NLP is a key part of AI. Tokenization is the first NLP step."
word_tokenize(text)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["playing", "played", "plays", "happily", "studies"]

stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer


lemmatizer = WordNetLemmatizer()
words = ["running", "cars", "better", "children", "feet"]

lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
print(lemmatized_words)
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
words = ["running", "cars", "better", "children", "feet"]
for word in words:
  print(f"Word: {word} | Stem: {stemmer.stem(word)} | Lemma: {lemmatizer.lemmatize(word)}")
```

```python
from nltk.stem import WordNetLemmatizer


lemmatizer = WordNetLemmatizer()
words = ["running", "cars", "better", "children", "feet"]

lemmatized_words = [lemmatizer.lemmatize(word,"v") for word in words]
print(lemmatized_words)
```
