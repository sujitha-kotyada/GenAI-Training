```python
import torch              #import pytorch and neural network module
import torch.nn as nn
```

```python
model=nn.Sequential(  #nn.sequential - stacks layers in order
    nn.Linear(2,4),   # 2 inputs,4 hidden neurons
    nn.ReLU(),        # adds non lonearity
    nn.Linear(4,1)    #single output neuron
)
```

INPUT AND TARGET DATA:
 this is XOR style data. x contains input features and y contains expected outputs

```python
x = torch.tensor([[0.,0.],[0.,1.],[1.,0.],[1.,1.]])
y = torch.tensor([[0.],[1.],[1.],[0.]])
```

LOSS FUNCTION AND OPTIMIZER : MSELoss measures prediction error  and Adam optimizer updates weights efficiently

```python
loss_fn = nn.MSELoss()
opt = torch.optim.Adam(model.parameters(),lr=0.01)
```

TRAINING LOOP

```python
for _ in range(1000):
  opt.zero_grad()
  loss=loss_fn(model(x),y)
  loss.backward()
  opt.step()
```

MODEL INFERENCE : We print the models predictions after training



```python
print(model(x))
```
